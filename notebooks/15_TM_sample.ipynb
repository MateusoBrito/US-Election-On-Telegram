```python
import pandas as pd

df_results = pd.read_csv("bertopic_parameter_tuning_results_sample_3.csv")
data_frame = pd.read_csv("preprocessed_english_titles.csv",engine = 'python', on_bad_lines='skip')
```

```python
# Calcula a média das métricas
df_results['mean_score'] = df_results[['coherence', 'diversity', 'silhouette']].mean(axis=1)

# 20 melhores por média
top20_mean = df_results.sort_values(by='mean_score', ascending=False).head(20)

# 20 melhores por diversidade, mas ordenados pela média
top20_diversity = (
    df_results.sort_values(by='diversity', ascending=False)
    .head(20)
    .sort_values(by='mean_score', ascending=False)
)

# 20 melhores por coerência, mas ordenados pela média
top20_coherence = (
    df_results.sort_values(by='coherence', ascending=False)
    .head(20)
    .sort_values(by='mean_score', ascending=False)
)

# 20 melhores por silhueta, mas ordenados pela média
top20_silhouette = (
    df_results.sort_values(by='silhouette', ascending=False)
    .head(20)
    .sort_values(by='mean_score', ascending=False)
)

best_mean_row = top20_mean
best_coherence_row = top20_coherence
best_silhouette_row = top20_silhouette
best_diversity_row = top20_diversity
# Exibe os resultados
display(best_mean_row)
display(best_coherence_row)
display(best_diversity_row)
display(best_silhouette_row)
```

```python
'''Remoção de Stopwords'''
from PreProcessing.pre_processing import PreProcessing
from nltk.corpus import stopwords

pp = PreProcessing(language="en")

custom_stopwords = [line.strip() for line in open('stopwords.txt', 'r', encoding='utf-8')]
english_stopwords = set(stopwords.words('english'))
pp.append_stopwords_list(list(english_stopwords - set(pp.stopwords)) + custom_stopwords)

data_frame["clean_text"] = data_frame["clean_text"].apply(
    lambda t: " ".join(
        [w for w in pp.remove_stopwords(t).split() if len(w) <= 20]
    )
)
```

```python
import csv

nan_count = data_frame['clean_text'].isna().sum()
df = data_frame[data_frame['clean_text'].notna()].reset_index(drop=True)
print(f"{nan_count} rows with NaN in 'clean_text' were removed.")

duplicate_count = df.duplicated(subset=['clean_text']).sum()
df.drop_duplicates(subset=['clean_text'], inplace=True)
print(f"{duplicate_count} duplicate rows based on 'clean_text' were removed.")

print(f"Total documents after cleaning: {len(df)}")

sample_df = df.sample(frac=0.1, random_state=42)
sample_docs = sample_df['clean_text'].tolist()
print(f"Sampled {len(sample_docs)} documents for topic modeling.")

print(len(df))
print(len(sample_df))
df.to_csv("processed_df.csv", index=False, quoting=csv.QUOTE_ALL)
sample_df.to_csv("sample_df.csv", index=False, quoting=csv.QUOTE_ALL)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from meu_bertopic import BERTopic
from bertopic.vectorizers import ClassTfidfTransformer
from umap import UMAP
from hdbscan import HDBSCAN

default_stopwords = []

def run_bertopic_modeling(docs, n_neighbors, n_components, min_dist, min_cluster_size, min_samples, cluster_eps,embedding_model_name="all-MiniLM-L6-v2"):
    umap_params = {
        "n_neighbors": n_neighbors, 
        "n_components": n_components, 
        "min_dist": min_dist,
        "metric":'cosine',
        "random_state":42
    }

    hdbscan_params = {
        "min_cluster_size": min_cluster_size, 
        "min_samples": min_samples,
        "prediction_data":True,
        "cluster_selection_epsilon": cluster_eps
    }

    bertopic_params = {
        'language': 'english',
        #'verbose': True,
        'top_n_words': 10
    }

    embedding_model = SentenceTransformer(embedding_model_name, device='cuda')
    vectorizer = TfidfVectorizer(stop_words=default_stopwords,ngram_range=(1, 1))

    umap_model = UMAP(**umap_params)
    hdbscan_model = HDBSCAN(**hdbscan_params)

    model = BERTopic(
        **bertopic_params,
        embedding_model=embedding_model,
        umap_model=umap_model,
        hdbscan_model=hdbscan_model,
        vectorizer_model=vectorizer,
        ctfidf_model=ClassTfidfTransformer(reduce_frequent_words=True)
    )

    topics, probs = model.fit_transform(docs)
    return model, topics, probs
```

```python
from IPython.display import display

def show_results(model):

    info = model.get_topic_info()
    display(info)

    topicos = [t for t in model.get_topics().keys() if t != -1]
    fig = model.visualize_barchart(topics=topicos, n_words=10)
    fig.show()
```

```python
import os

def save_results(model, dir):
    json_path = os.path.join(dir, "topics.json")
    model.save_json(json_path) 

    topicos = [t for t in model.get_topics().keys() if t != -1]
    fig = model.visualize_barchart(topics=topicos, n_words=10)

    barchart_path = os.path.join(dir, "barchart.html")
    fig.write_html(barchart_path) 

    model_path = os.path.join(dir, "bertopic_model")
    model.save(model_path)

    topics_info = model.get_topic_info()
    csv_path = os.path.join(dir, "topics_info.csv")
    topics_info.to_csv(csv_path, index=False)
    
    print(f"Modelo e resultados salvos no diretório '{dir}' (gráfico salvo como HTML)")
```

```python
for i,row in enumerate(best_mean_row.itertuples()):
    print("Running BERTopic with parameters for best mean score:")
    print(row)
    model, topics, probs = run_bertopic_modeling(
        docs=sample_docs,
        n_neighbors=row.n_neighbors,
        n_components=row.n_components,
        min_dist=row.min_dist,
        min_cluster_size=row.min_cluster_size,
        min_samples=row.min_samples,
        cluster_eps=row.cluster_eps
    )
    dir = f"best_mean/{i}"
    if not os.path.exists(dir):
        os.makedirs(dir)
    save_results(model, dir)
```
