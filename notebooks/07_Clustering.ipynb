This code aims to cluster the titles of YouTube videos, trying to search for videos related to politics

```python
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.corpus import stopwords
import nltk

nltk.download("stopwords")
stop_words = stopwords.words("english") + stopwords.words("portuguese") + stopwords.words("french")
```

```python
df = pd.read_csv("../data/youtube_titles_2.csv")
display(df)
```

```python
def pre_processing(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-ZÀ-ÿ\s]", "", text)
    return text.strip()

df["clean_title"] = df["title"].apply(pre_processing)
display(df)
```
