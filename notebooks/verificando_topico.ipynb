```python
from collections import Counter
import re

# juntar todo o texto do df_crime2
all_text = " ".join(df_crime2["text"].astype(str))

# tokenizar (somente letras, tudo minúsculo)
tokens = re.findall(r"\b[a-zA-Z]+\b", all_text.lower())

# remover stopwords básicas
stopwords = {"the","and","to","of","in","a","is","that","for","it","on","with","as","at","this","was","be","by","are","an","from"}

tokens_clean = [t for t in tokens if t not in stopwords]

# contar
freq = Counter(tokens_clean)

# ver as 30 mais comuns
freq.most_common(30)
```
