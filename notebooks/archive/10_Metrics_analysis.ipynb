### **1. Correlation of sharing with views**
___

```python
file_path = "data/youtube_links.jsonl"
```

#### **1.1 Sharing count per video**
____

```python
import pandas as pd
import gzip
import json
from tqdm.notebook import tqdm

dados_processados = []

try:
    with open(file_path, "rt", encoding="utf-8") as f:
        for line in tqdm(f, desc="Processando URLs"):
            try:
                data = json.loads(line)

                url = data.get('url', None)
                occurrences = data.get('occurrences', [])
            
                dados_processados.append({
                    'url': url,
                    'sharing_count': len(occurrences)
                })

            except json.JSONDecodeError:
                # Se uma linha não for um JSON válido, ela é ignorada
                print(f"Aviso: Linha com JSON inválido foi ignorada.")
                continue
            except Exception as e:
                # Captura outros erros inesperados na linha
                print(f"Ocorreu um erro ao processar uma linha: {e}")
                continue
except FileNotFoundError:
    print(f"ERRO: O arquivo '{file_path}' não foi encontrado.")
except Exception as e:
    # Captura outros erros inesperados (ex: arquivo corrompido)
    print(f"Ocorreu um erro geral: {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import random

df = pd.DataFrame(dados_processados)
media_compartilhamentos = df['sharing_count'].mean()
print("--- Estatísticas Gerais ---")
print(f"Média de compartilhamentos por vídeo: {media_compartilhamentos:.2f}")
limites_dos_bins = [-1, 0, 1, 5, 10, 50, 100, 500, 1000, 5000, 20000, np.inf]

nomes_dos_bins = [
    '0',
    '1',
    '2 a 5',
    '6 a 10',
    '11 a 50',
    '51 a 100',
    '101 a 500',
    '501 a 1.000',
    '1.001 a 5.000',
    '5.001 a 20.000',
    'Mais de 20.000'
]

df['faixa_compartilhamento'] = pd.cut(
    df['sharing_count'],
    bins=limites_dos_bins,
    labels=nomes_dos_bins
)

contagem_por_faixa = df['faixa_compartilhamento'].value_counts().sort_index()
print("\n--- Contagem de Vídeos por Faixa ---")
print(contagem_por_faixa)
```

```python
media_views = df['view_count'].mean()
mediana_views = df['view_count'].median()
desvio_views = df['view_count'].std()
min_views = df['view_count'].min()
max_views = df['view_count'].max()
q1 = df['view_count'].quantile(0.25)
q3 = df['view_count'].quantile(0.75)
iqr = q3 - q1  # Intervalo interquartil

print(f"Média de visualizações: {media_views:,.2f}")
print(f"Mediana de visualizações: {mediana_views:,.2f}")
print(f"Desvio padrão: {desvio_views:,.2f}")
print(f"Valor mínimo: {min_views:,.0f}")
print(f"1º quartil (Q1): {q1:,.0f}")
print(f"3º quartil (Q3): {q3:,.0f}")
print(f"Intervalo interquartil (IQR): {iqr:,.0f}")
print(f"Valor máximo: {max_views:,.0f}")
```

### **2. How many groups one videos is shared**
___

```python
import pandas as pd
import gzip
import json
from tqdm.notebook import tqdm

dados_processados = []
i =0
try:
    with open(file_path, "rt", encoding="utf-8") as f:
        for line in tqdm(f, desc="Processando URLs"):
            try:
                data = json.loads(line)

                url = data.get('url', None)
                occurrences = data.get('occurrences', [])

                # Obter todos os grupos únicos
                unique_groups = set([occ['folder'] for occ in occurrences])
                num_unique_groups = len(unique_groups)
            
                dados_processados.append({
                    'url': url,
                    'unique_group_count': num_unique_groups
                })
                
            except json.JSONDecodeError:
                # Se uma linha não for um JSON válido, ela é ignorada
                print(f"Aviso: Linha com JSON inválido foi ignorada.")
                continue
            except Exception as e:
                # Captura outros erros inesperados na linha
                print(f"Ocorreu um erro ao processar uma linha: {e}")
                continue
except FileNotFoundError:
    print(f"ERRO: O arquivo '{file_path}' não foi encontrado.")
except Exception as e:
    # Captura outros erros inesperados (ex: arquivo corrompido)
    print(f"Ocorreu um erro geral: {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import random

df = pd.DataFrame(dados_processados)
media_compartilhamentos = df['unique_group_count'].mean()
print("--- Estatísticas Gerais ---")
print(f"Média de compartilhamentos por vídeo: {media_compartilhamentos:.2f}")
limites_dos_bins = [-1, 0, 1, 5, 10, 50, 100, 500, 1000, 5000, 20000, np.inf]

nomes_dos_bins = [
    '0',
    '1',
    '2 a 5',
    '6 a 10',
    '11 a 50',
    '51 a 100',
    '101 a 500',
    '501 a 1.000',
    '1.001 a 5.000',
    '5.001 a 20.000',
    'Mais de 20.000'
]

df['faixa_compartilhamento'] = pd.cut(
    df['unique_group_count'],
    bins=limites_dos_bins,
    labels=nomes_dos_bins
)

contagem_por_faixa = df['faixa_compartilhamento'].value_counts().sort_index()
print("\n--- Contagem de Vídeos por Faixa ---")
print(contagem_por_faixa)
```
