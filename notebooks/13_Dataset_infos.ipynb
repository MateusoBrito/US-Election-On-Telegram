```python
import pandas as pd
```

```python
df1 = pd.read_csv("../data/youtube_titulos_saida.csv")
display(df1)
```

```python
df2 = pd.read_csv("../data/preprocessed_english_titles.csv")
display(df2)
```

```python
print("\nJuntando os DataFrames para encontrar os vídeos em comum...")
merged_df = pd.merge(df1, df2, on='video_id', how='inner')
print(f"Encontrados {len(merged_df)} vídeos em comum.")

print("Selecionando e renomeando as colunas desejadas...")

colunas_originais = [
    'video_id',
    'view_count',
    'like_count',
    'comment_count'
]

try:
    final_df = merged_df[colunas_originais].copy() 
except KeyError as e:
    print(f"\nERRO: Uma das colunas não foi encontrada no DataFrame! Detalhe: {e}")
    print("Verifique se os nomes em 'colunas_originais' estão corretos.")
    exit()


display(final_df)

final_df.to_csv("../data/df_analise", index=False)
```
